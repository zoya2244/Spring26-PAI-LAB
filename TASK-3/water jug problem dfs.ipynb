```python
jug1_capacity = 4
jug2_capacity = 3
target = 2

visited = set()


def dfs(state, path):
    x, y = state

    if x == target or y == target:
        print("Water Jug Problem\n")
        print("Steps:")
        for step in path:
            print(step)
        return True

    visited.add(state)

    x_cap = jug1_capacity
    y_cap = jug2_capacity

    next_states = []

    next_states.append(((x_cap, y), "Fill Jug1"))
    next_states.append(((x, y_cap), "Fill Jug2"))
    next_states.append(((0, y), "Empty Jug1"))
    next_states.append(((x, 0), "Empty Jug2"))
    pour = min(x, y_cap - y)
    next_states.append(((x - pour, y + pour), "Pour Jug1 > Jug2"))

    pour = min(y, x_cap - x)
    next_states.append(((x + pour, y - pour), "Pour Jug2 > Jug1"))

    for new_state, action in next_states:
        if new_state not in visited:
            if dfs(new_state, path + [f"{action} -> {new_state}"]):
                return True

    return False


dfs((0, 0), [])
```
